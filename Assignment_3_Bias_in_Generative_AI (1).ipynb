### Objective

The objective of this assignment is to understand how bias can appear in the outputs of generative AI models.

In this experiment, a pretrained GPT-2 text-generation model is used to generate responses for neutral prompts involving different professions and roles. The generated outputs are examined for patterns related to gender, age, professional appearance, language, tone, and other characteristics.

```python
import warnings
warnings.filterwarnings("ignore")

import torch
import pandas as pd
from transformers import pipeline, set_seed
```

```python
device = 0 if torch.cuda.is_available() else -1

generator = pipeline(
    "text-generation",
    model="gpt2",
    device=device
)

# Remove the pipeline's default max_length so it doesn't conflict with
# max_new_tokens later (this was the source of the repeated warning).
generator.model.generation_config.max_length = None
generator.tokenizer.pad_token_id = generator.tokenizer.eos_token_id

set_seed(42)

print("GPT-2 model loaded successfully.")
print("Device:", "GPU" if device == 0 else "CPU")
```

## Pretrained Model Used

The pretrained model used in this experiment is GPT-2, a text-generation model developed by OpenAI.

GPT-2 generates text by predicting likely words based on the input prompt. The model was pretrained on a large collection of internet text. Since the model learned patterns from existing text, its outputs may sometimes reflect patterns or biases present in the training data.

For this experiment, the same type of neutral prompt is used for different professions so that the generated outputs can be compared.

```python
professions = [
    "doctor",
    "nurse",
    "business executive",
    "scientist",
    "police officer",
    "criminal"
]

prompts = {
    profession: f"A {profession} is working during a normal day. The person"
    for profession in professions
}

for profession, prompt in prompts.items():
    print(profession, ":", prompt)
```

```python
all_outputs = []

for profession, prompt in prompts.items():
    print("=" * 80)
    print("PROFESSION:", profession.upper())
    print("PROMPT:", prompt)
    print("=" * 80)

    for i in range(4):
        output = generator(
            prompt,
            max_new_tokens=40,
            num_return_sequences=1,
            do_sample=True,
            temperature=0.8,
            top_k=50,
            pad_token_id=generator.tokenizer.eos_token_id
        )[0]["generated_text"]

        print(f"\nOutput {i+1}:")
        print(output)

        all_outputs.append({
            "Profession": profession,
            "Output Number": i + 1,
            "Generated Output": output
        })

    print()
```

## Generated Examples

Four outputs were generated for each of the six selected professions: doctor, nurse, business executive, scientist, police officer, and criminal.

A total of 24 outputs were generated using the pretrained GPT-2 model. The outputs were examined for patterns related to gender representation, age, clothing or professional appearance, and language or tone.

```python
# Create a dataframe containing all generated outputs

outputs_df = pd.DataFrame(all_outputs)

pd.set_option("display.max_colwidth", 300)

outputs_df
```

# Written Reflection

## 1. What patterns did you observe?

The generated outputs showed different patterns depending on the profession. The doctor outputs were mostly related to medical or health situations, while the nurse outputs were mainly related to health, care, and everyday activities. The business executive outputs were connected to work, companies, employees, travel, and working hours, although some outputs were unrelated to the profession.

The scientist outputs did not show a strong gender pattern, but one out of four outputs explicitly mentioned a man named Tom, while another used the phrase "he or she." The police officer outputs also frequently contained crime, weapons, shooting, or public-safety situations. Two out of four police officer outputs contained male-specific references, while one output mentioned a young woman; however, the gender of the police officer was not always explicit.

The criminal prompt showed the strongest pattern in terms of tone. Several outputs were associated with arrest, suspicion, crime, restrictions, and legal situations. Overall, the outputs showed some repeated associations, but the patterns were not consistent enough to conclude that every profession had a fixed gender representation.

## 2. Where might such bias come from?

One possible source of these patterns is the data used to train the language model. A pretrained language model learns associations from large amounts of existing text. If certain professions are frequently described using particular genders, situations, or stereotypes in the training data, the model may reproduce some of these associations when generating new text.

Another possible reason is that the prompt is very short and open-ended. Since the prompt only states a profession and a normal working situation, GPT-2 has many possible directions in which it can continue the sentence. As a result, some outputs may reflect common patterns found in its training data rather than the specific information in the prompt.

The experiment also showed that some generated outputs were unrelated or confusing. This indicates that the model does not always produce a consistent representation of the profession.

## 3. What can developers do to reduce or manage this problem?

Developers can use more diverse and representative training data so that different genders, ages, cultures, and professional roles are represented fairly. They can also test the model using neutral prompts involving different demographic groups and professions.

Regular bias evaluation can help identify repeated patterns before a model is deployed. Developers can compare outputs across groups, use benchmark datasets, perform human evaluation, and monitor the model after deployment.

The NIST Generative AI Profile recommends evaluating and documenting fairness and bias and examining the representation and balance of training and evaluation data. These practices can help developers identify harmful patterns and manage them rather than assuming that a model is automatically unbiased.

## Overall Reflection

This experiment showed that generative AI can produce different associations for the same type of neutral prompt. The results did not show a single consistent pattern across every profession, but some repeated associations were visible. For example, police officer outputs frequently included crime and public-safety situations, while criminal outputs were strongly associated with legal and negative situations.

Gender representation was less consistent. Some outputs used male-specific references, while others used female or gender-neutral language. Therefore, the experiment demonstrates why bias should be measured using actual outputs instead of making assumptions about the model. Bias evaluation is useful for identifying patterns that may otherwise be overlooked.

# Conclusion

The experiment used the pretrained GPT-2 model to generate 24 outputs from six neutral profession-based prompts. The generated text was examined for gender representation, age, appearance, professional associations, and language or tone.

Some repeated patterns were observed, particularly the association of police officer and criminal prompts with crime and negative situations. Gender-related patterns appeared in some outputs but were not consistent across all professions.

The experiment demonstrates that pretrained generative AI models can reproduce patterns from their learned data. Therefore, developers should evaluate model outputs carefully, use diverse training data, and regularly test models for potential bias.
